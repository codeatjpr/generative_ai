```python
from langchain_core.tools import tool
from langchain.agents import create_agent

@tool
def book_movie(
    username: str,
    movie_name: str,
    theatre: str,
    seats: int
) -> dict:
    """Create a movie ticket booking request."""
    # Demo only: real app would call a booking API.
    return {
        "booking_id": "MOV-1234",
        "username": username,
        "movie": movie_name,
        "theatre": theatre,
        "seats": seats,
        "status": "booking_created"
    }

@tool
def phone_pay_payment(
    user_id: str,
    amount: float,
    phone_number: str
) -> dict:
    """Initiate a demo payment request. Never use for real payments."""
    # Mock response only — no money is transferred.
    return {
        "status": "success_demo",
        "txn_id": "TXN-5678",
        "amount_paid": amount
    }

model = "gpt-4o-mini"
tools = [book_movie, phone_pay_payment]
agent = create_agent(model, tools=tools)

query = """
My name is Raj. I want 3 tickets for Example Movie
at Example Cinema. My user ID is U123.
The amount is Rs 900 and my phone number is 9000000000.
"""

result = agent.invoke({
    "messages": [{"role": "user", "content": query}]
})

print(result["messages"][-1].content)
```
