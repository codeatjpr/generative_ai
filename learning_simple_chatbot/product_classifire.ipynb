```python

import os
from dotenv import load_dotenv

from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate

load_dotenv()

# Initialize LLM
llm = ChatGroq(
    model="openai/gpt-oss-120b"
)

# Product categories
categories = [
    "Clothing",
    "Footwear",
    "Accessories",
    "Electronics",
    "Home & Kitchen"
]

# Products to classify
products = [
    "Men's Cotton T-Shirt",
    "Running Shoes",
    "Leather Belt",
    "Bluetooth Headphones",
    "Non-stick Frying Pan",
    "Women's Denim Jeans",
    "Smart Watch",
    "Leather Handbag"
]

# Common instruction
instruction = f"""
You are a retail product classification assistant.

Classify the given product into exactly one category.

Available categories:
{categories}

Rules:
- Return only the category name.
- Do not provide explanations.
- Never create a new category.
"""

# ZERO-SHOT PROMPT
zero_shot_prompt = ChatPromptTemplate.from_messages([
    ("system", instruction),
    ("human", "Classify this product: {product}")
])

# ONE-SHOT PROMPT
one_shot_prompt = ChatPromptTemplate.from_messages([
    ("system", instruction),
    ("human", """
Example:
Product: Men's Formal Shirt
Category: Clothing

Now classify:
Product: {product}
Category:
""")
])

# FEW-SHOT PROMPT
few_shot_prompt = ChatPromptTemplate.from_messages([
    ("system", instruction),
    ("human", """
Examples:

Product: Men's Formal Shirt
Category: Clothing

Product: Sports Sneakers
Category: Footwear

Product: Leather Wallet
Category: Accessories

Product: Wireless Mouse
Category: Electronics

Product: Coffee Maker
Category: Home & Kitchen

Now classify:
Product: {product}
Category:
""")
])

# Create chains
zero_shot_chain = zero_shot_prompt | llm
one_shot_chain = one_shot_prompt | llm
few_shot_chain = few_shot_prompt | llm



# Run classifier
def classify_products(chain, products):

    results = []
   

    for product in products:

        response = chain.invoke({
            "product": product
        })

    

        results.append({
            "product": product,
            "category": response.content.strip()
        })

    return results

# Execute all three approaches
zero_results, zero_tokens = classify_products(
    zero_shot_chain, products
)

one_results, one_tokens = classify_products(
    one_shot_chain, products
)

few_results, few_tokens = classify_products(
    few_shot_chain, products
)

# Display results
for name, results, usage in [
    ("ZERO-SHOT", zero_results, zero_tokens),
    ("ONE-SHOT", one_results, one_tokens),
    ("FEW-SHOT", few_results, few_tokens)
]:

    print(f"\n========== {name} ==========")

    for result in results:
        print(
            result["product"],
            "->",
            result["category"]
        )

```
