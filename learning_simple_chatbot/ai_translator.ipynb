## AI Translator Multilingual 

```python
import os
from dotenv import load_dotenv

from groq import Groq
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
```

```python
load_dotenv()

api_key = os.getenv("GROQ_API_KEY")
#api_key = os.getenv("OPEN_API_KEY")  # This line will overwrite the previous api_key value
if not api_key:
    raise ValueError("GROQ_API_KEY not found!")

```

```python
llm = ChatGroq(
    model="openai/gpt-oss-120b"
)
```

```python
response = llm.invoke(
    "Hello, how are you?."
)

print(response.content)
```

```python
# 3. Create ChatPromptTemplate
prompt = ChatPromptTemplate.from_messages([

    (
        "system",
        """
        You are a professional multilingual translator.

        Translate the given text into the requested target language.

        Instructions:
        - Preserve the original meaning.
        - Use natural and grammatically correct language.
        - Maintain the original tone.
        - Do not add explanations.
        - Return only the translated text.
        """
    ),

    (
        "human",
        """
        Text to translate:
        {input_text}

        Target language:
        {target_language}
        """
    )

])
```

```python
chain = prompt | llm
```

```python
response = chain.invoke({
    "input_text":"Hello, how are you?",
    "target_language":"hindhi"
}
)

print(response.content)
```
