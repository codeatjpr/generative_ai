# Structured output with JSON Schema

Pass the `BankRequest` JSON Schema to `with_structured_output` to describe the fields the model must return. The result is a regular Python dictionary.

This example uses `gpt-5.4-mini`. Set `OPENAI_API_KEY` in your environment or repository `.env` file before running it.

```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI

load_dotenv()

bank_request_schema = {
    "title": "BankRequest",
    "type": "object",
    "properties": {
        "intent": {"type": "string"},
        "department": {"type": "string"},
        "account_last4": {"type": ["string", "null"]},
        "priority": {"type": "string"},
        "needs_human": {"type": "boolean"},
    },
    "required": [
        "intent",
        "department",
        "account_last4",
        "priority",
        "needs_human",
    ],
    "additionalProperties": False,
}

model = ChatOpenAI(model="gpt-5.4-mini", temperature=0)
model_with_structure = model.with_structured_output(
    bank_request_schema,
    method="json_schema",
)

customer_message = (
    "My debit card is blocked and I need it unblocked. "
    "The last four digits are 1234."
)
response = model_with_structure.invoke(
    "Classify this customer request using the BankRequest schema. "
    "Set account_last4 to null if no last four digits are supplied. "
    f"Customer request: {customer_message}"
)
print(response)
print(
    f"Intent: {response['intent']}; department: {response['department']}; "
    f"last four: {response['account_last4']}; priority: {response['priority']}; "
    f"needs human: {response['needs_human']}"
)
```
