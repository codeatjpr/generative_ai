# Structured output with `TypedDict`

Define the fields a bank help desk needs with `TypedDict` and `Literal` types, then pass the type to `with_structured_output`. LangChain returns the structured result as a Python dictionary.

This example uses `gpt-5.4-mini`. Set `OPENAI_API_KEY` in your environment or repository `.env` file before running it.

```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from typing import TypedDict, Literal

load_dotenv()


class BankRequestDict(TypedDict):
    """Classified customer request for bank help desk."""

    intent: Literal["CARD_UNBLOCK", "CARD_BLOCK", "OTHER"]
    department: Literal["CARDS", "GENERAL"]
    account_last4: str | None
    priority: Literal["LOW", "MEDIUM", "HIGH"]
    needs_human: bool


model = ChatOpenAI(model="gpt-5.4-mini", temperature=0)
model_with_structure = model.with_structured_output(BankRequestDict)

customer_message = (
    "My debit card is blocked and I need it unblocked. "
    "The last four digits are 1234."
)
response = model_with_structure.invoke(
    "Classify this customer request using the BankRequestDict schema. "
    "Set account_last4 to null if no last four digits are supplied. "
    f"Customer request: {customer_message}"
)
print(response)
print(
    f"Intent: {response['intent']}; department: {response['department']}; "
    f"last four: {response['account_last4']}; priority: {response['priority']}; "
    f"needs human: {response['needs_human']}"
)
```
